```python
my_dict = {'k1':1, 'k2':2 , 'k3':3}

for i in my_dict:
    print(i)
#iterating loop in dictionary
```

```python
for i in my_dict.items():
    print(i)
 # printing out items in dictionary iterating   
```

```python
# how do you unpack the values from the dictionary?

for  a,b in my_dict.items():
    print(b)
# unpacking value in dictionary. 
```

 ## While Loops

```python
i = 0 
while i < 5:
  print(i)
  i = i + 1
  #i += 1
```

```python
i = 0 
while i < 5:
  print(i)
  i = i + 1
  #i += 1
else:
  print("i is not less than 5")
```
